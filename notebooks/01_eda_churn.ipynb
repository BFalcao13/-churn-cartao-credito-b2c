```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/raw/BankChurners.csv')

df.head()
```

```python
df.info()
```

```python
df.columns.tolist()
```

```python
colunas_naive_bayes = [col for col in df.columns if col.startswith('Naive_Bayes')]
print(colunas_naive_bayes)
```

```python
df = df.drop(columns=colunas_naive_bayes, errors='ignore')

df.shape
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Customer_Age', hue='Attrition_Flag', kde=True, multiple='layer', alpha=0.5)
plt.title('Distribuição de Idade por Status de Churn')
plt.xlabel('Idade do Cliente')
plt.ylabel('Quantidade de Clientes')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Total_Trans_Ct', hue='Attrition_Flag', kde=True, multiple='layer', alpha=0.5)
plt.title('Distribuição de Idade por Status de Churn')
plt.xlabel('Idade do Cliente')
plt.ylabel('Quantidade de Clientes')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x='Months_Inactive_12_mon', hue='Attrition_Flag', kde=True, multiple='layer', alpha=0.5, discrete=True)
plt.title('Meses de Inatividade (12 meses) por Status de Churn')
plt.xlabel('Meses Inativos')
plt.ylabel('Quantidade de Clientes')
plt.show()
```

```python
taxa_por_inatividade = df.groupby('Months_Inactive_12_mon')['Attrition_Flag'].apply(
    lambda x: (x == 'Attrited Customer').mean() * 100
).round(2)

taxa_por_inatividade
```
